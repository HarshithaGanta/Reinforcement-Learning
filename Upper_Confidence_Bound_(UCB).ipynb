### **Upper Confidence Bound for Multi-Armed Bandit**

```python
import math

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Part B – Initialize N, R, UCB and Q tables
N = {'A': 0, 'B': 0, 'C': 0}       # Number of times each machine is played
R = {'A': 0, 'B': 0, 'C': 0}       # Total reward received from each machine
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0} # Estimated Q-value
UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0}

# Reward patterns given in the assignment
reward_pattern = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

reward_index = {'A': 0, 'B': 0, 'C': 0}

total_reward = 0

# Part C, D and E – UCB selection, reward collection and Q update
for play in range(1, N_play + 1):

    # Calculate UCB for every machine
    for machine in machines:

        # Unplayed machines get infinity so that
        # every machine is tried at least once
        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    # Select machine having highest UCB
    selected_machine = max(machines, key=lambda x: UCB[x])

    print("\nPlay", play)
    print("Selected Machine:", selected_machine)

    # For the suggested experiment, use the given reward pattern
    index = reward_index[selected_machine] % len(
        reward_pattern[selected_machine]
    )

    reward = reward_pattern[selected_machine][index]
    reward_index[selected_machine] += 1

    print("Reward obtained:", reward)

    # Update number of times selected machine was played
    N[selected_machine] += 1

    # Update total reward
    R[selected_machine] += reward
    total_reward += reward

    # Incremental sample-average Q-value update
    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]

    print("Updated Q-value:", round(Q[selected_machine], 4))


# Part F – Display final results
print("\n" + "=" * 60)
print("FINAL RESULTS")
print("=" * 60)

print("\nMachine\tN\tR\tUCB\t\tQ-value")
print("-" * 60)

for machine in machines:

    # Calculate final UCB value
    if N[machine] == 0:
        UCB[machine] = float('inf')
    else:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(N_play)) / N[machine]
        )

    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]}\t"
        f"{UCB[machine]:.4f}\t\t"
        f"{Q[machine]:.4f}"
    )

print("-" * 60)
print("Total Reward Received =", total_reward)
```
