### **Calculate Undiscounted and Discounted Returns**

```python
-n=int(input("Enter number of time steps: "))
rewards=[]

for i in range(n):
    r=float(input(f"Enter reward at time step {i+1}: "))
    rewards.append(r)

gamma=float(input("Enter discount factor γ: "))

undiscounted=sum(rewards)
discounted=sum(rewards[i]*(gamma**i) for i in range(n))

print("\nEntered reward values:",rewards)
print("Undiscounted return:",undiscounted)
print("Discounted return:",discounted)
```

**Create grid_env_det.py**

```python
%%writefile grid_env_det.py

states=["S0","S1","S3","S4","G"]
actions=["Up","Down","Left","Right"]
gamma=0.9

transitions={
    "S0":{"Up":"S0","Down":"S0","Left":"S0","Right":"S1"},
    "S1":{"Up":"S1","Down":"S3","Left":"S0","Right":"G"},
    "S3":{"Up":"S1","Down":"S3","Left":"S3","Right":"S4"},
    "S4":{"Up":"G","Down":"S4","Left":"S3","Right":"S4"},
    "G":{"Up":"G","Down":"G","Left":"G","Right":"G"}
}

def transition(state,action):
    return transitions[state][action]

def reward(state,action):
    next_state=transition(state,action)
    if next_state=="G":
        return 10
    return -1

def step(state,action):
    next_state=transition(state,action)
    r=reward(state,action)
    done=next_state=="G"
    return next_state,r,done
```

**User Input for Current State and Action**


```python
%%writefile main.py

from grid_env_det import states,actions,gamma,transition,reward,step

print("State Space:",states)
print("Action Space:",actions)
print("Discount Factor:",gamma)

print("\nTransition Function:")
for state in states:
    for action in actions:
        print(state,action,"->",transition(state,action))

print("\nReward Example:")
print("S0 + Right =",reward("S0","Right"))

print("\nNext State:")
print("Current State = S1")
print("Action = Right")
print("Next State =",transition("S1","Right"))
```

**Display Complete MDP**

```python
%%writefile main.py

from grid_env_det import states,actions,gamma,transition,reward,step

print("STATE SPACE")
print(states)

print("\nACTION SPACE")
print(actions)

print("\nREWARD FUNCTION")
print("Normal Movement = -1")
print("Goal = +10")
print("Obstacle = -5")

print("\nDISCOUNT FACTOR")
print(gamma)

print("\nTRANSITION FUNCTION")
for s in states:
    for a in actions:
        print(f"{s} + {a} -> {transition(s,a)}")

state=input("\nEnter current state: ")
action=input("Enter action: ")

next_state,r,done=step(state,action)

print("\nRESULT")
print("Current State:",state)
print("Action:",action)
print("Next State:",next_state)
print("Reward:",r)
print("Episode Terminated:",done)
```
