```python
# Part A - Machine Initialization

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

print("Machines:", machines)
print("Number of machines:", M)
print("Number of plays:", N_play)
print("Exploration probability:", epsilon)
```

```python
# Part B - Initialize knowledge tables

N = [0, 0, 0]       # Number of times each machine is played
R = [0, 0, 0]       # Total reward obtained from each machine
Q = [0.0, 0.0, 0.0] # Estimated Q-value of each machine

print("Initial N:", N)
print("Initial R:", R)
print("Initial Q:", Q)
```

```python
# Parts C, D and E - Modified Epsilon-Greedy Strategy

import random

# Store history for final display
history = []

total_reward = 0

for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Find machine with highest Q-value
    highest_Q = max(Q)
    best_machine = Q.index(highest_Q)

    # Modified epsilon-greedy decision
    if r < epsilon:
        # Exploration:
        # Select randomly from machines except highest Q machine
        possible_machines = [i for i in range(M) if i != best_machine]
        selected = random.choice(possible_machines)
        decision = "Exploration"
    else:
        # Exploitation:
        # Select machine with highest Q-value
        selected = best_machine
        decision = "Exploitation"

    print("\n--------------------------------")
    print("Play", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Selected machine:", machines[selected])

    # Get reward from user
    reward = float(input("Enter reward: "))

    # Update number of plays
    N[selected] += 1

    # Update total reward
    R[selected] += reward
    total_reward += reward

    # Incremental sample-average Q-value update
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    # Store play information
    history.append([
        play,
        machines[selected],
        reward,
        N[selected],
        R[selected],
        Q[selected]
    ])

    print("Updated N:", N)
    print("Updated R:", R)
    print("Updated Q:", [round(q, 2) for q in Q])
```
